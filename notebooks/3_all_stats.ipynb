# 3. Every stat, per game against per possible game

The same idea as notebook 2, for six stats: points, assists, rebounds, steals, blocks and **stocks** (steals + blocks). Each stat is counted per game played and per possible game (every game his team played while he was on the roster).

Whole league, no minimum games. 2026-27 is still being played.

```python
import matplotlib.pyplot as plt
import pandas as pd

from xppg import data, plotting
from xppg.raw import PROJECT_ROOT

plotting.use_style()
ps = pd.read_parquet(PROJECT_ROOT / "data" / "processed" / "player_seasons.parquet")
STATS = ["pts", "ast", "reb", "stl", "blk", "stk"]
```

## 1. Six stats, seven seasons

**What you're looking at:** one row per season and one block per stat. In each block, the left list is the top 10 per game played and the right list is the top 10 per possible game. Between them, an arrow shows where each left-hand player ranks per possible game: **blue ↗** he rose, **red ↘** he fell, **=** no change. The number beside the arrow is his new rank.  
**Why it matters:** a big red fall means a player's per-game number hid a lot of missed games. Some left-hand players played only a game or two (there is no minimum), so they fall hundreds of places: one good night is not a season.

```python
fig = plotting.rank_grid(ps, data.SEASONS, STATS, in_progress=data.IN_PROGRESS,
                         title="Per game against per possible game, 2020-21 to 2026-27")
plotting.save(fig, "3_rank_grid")
plt.show()
```

## 2. Each season, closer up

**What you're looking at:** the same chart, one season at a time, so the names are easier to read.  
**Why it matters:** the full grid is large; this is the same information at a readable size.

```python
for season in data.SEASONS:
    fig = plotting.rank_grid(ps, [season], STATS, in_progress=data.IN_PROGRESS)
    plt.show()
```
